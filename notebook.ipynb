```python
# Check for duplicate rows in the DataFrame
duplicates = df.duplicated().sum()

print("Number of duplicate rows:", duplicates)
```

```python
# Drop rows with missing values
df = df.dropna()
```

```python
# Check for missing values in the DataFrame
df.isnull().sum()
```

```python
## Filtering & Subsetting: Filter invalid records with physically impossible negative values (Speed or Distance < 0)

# فحص القيم السالبة المستحيلة في المسافة والسرعة
invalid_records = df[(df['Distance_Travelled_km'] < 0) | (df['Speed_kmh'] < 0)]
invalid_records[['Vehicle_ID', 'Speed_kmh', 'Distance_Travelled_km']]
```

```python
# Decision: Drop rows with negative distance or speed values
# Reason: Distance and speed cannot be negative; these represent corrupt sensor recordings.
df = df[(df['Distance_Travelled_km'] >= 0) & (df['Speed_kmh'] >= 0)]

print("Total rows after filtering invalid records:", len(df))
```

```python
# فحص القيم الشاذة في السرعة وضغط الإطارات والرطوبة
outliers = df[(df['Speed_kmh'] > 120) | (df['Tire_Pressure_psi'] > 35) | (df['Humidity_%'] > 100)]
outliers[['Vehicle_ID', 'Speed_kmh', 'Tire_Pressure_psi', 'Humidity_%']]
```

```python
# Decision: Cap extreme outliers (Speed > 120 km/h, Tire Pressure > 35 psi)
# Reason: These spikes represent sensor errors rather than real vehicle dynamics. 
# Capping preserves valid trip records without distorting statistical distributions.

#تعديل القيم الشاذة لتستقر عند الحدود الطبيعية المقبولة (120 للسرعة و 35 للإطارات)
df['Speed_kmh'] = df['Speed_kmh'].clip(upper=120.0)
df['Tire_Pressure_psi'] = df['Tire_Pressure_psi'].clip(upper=35.0)

# طباعة أقصى القيم بعد التحجيم للتأكد من انضباطها
print("Max Speed after capping:", df['Speed_kmh'].max())
print("Max Tire Pressure after capping:", df['Tire_Pressure_psi'].max())
```

```python
# التحقق النهائي من عدم وجود قيم مفقودة أو تكرار
print("Final Dataset Shape:", df.shape)
print("Remaining Missing Values:", df.isnull().sum().sum())
print("Remaining Duplicates:", df.duplicated().sum())

# حفظ الملف النظيف النهائي
df.to_csv("clean_ev_energy_consumption.csv", index=False)
print("Clean dataset saved successfully as 'clean_ev_energy_consumption.csv'!")
```

```python
# Grouping & Sorting: حساب متوسط استهلاك الطاقة لكل نمط قيادة وترتيب النتائج تنازلياً
df.groupby('Driving_Mode')['Energy_Consumption_kWh'].mean().sort_values(ascending=False)
```

```python
# 1. إنشاء جدول الربط
mode_name = pd.DataFrame({
    'Road_Type': [1, 2, 3],
    'Mode': ['Highway', 'Urban', 'Rural']
})

# 2. التأكد من تطابق نوع البيانات في كلا الجدولين (تحويلهما إلى int)
df['Road_Type'] = df['Road_Type'].astype(int)
mode_name['Road_Type'] = mode_name['Road_Type'].astype(int)

# 3. الدمج وعرض النتيجة
df = pd.merge(df, mode_name, on='Road_Type', how='left')
df[['Road_Type', 'Mode', 'Energy_Consumption_kWh']].head()
```

```python
# Merging: إنشاء جدول تفسيري ودمجه مع البيانات لتوضيح أسماء أنماط القيادة
mode_names = pd.DataFrame({
    'Driving_Mode': [1, 2, 3],
    'Mode_Name': ['Eco', 'Normal', 'Sport']
})
df = pd.merge(df, mode_names, on='Driving_Mode', how='left')
df[['Driving_Mode', 'Mode_Name', 'Energy_Consumption_kWh']].head()
```

```python
# حساب متوسط استهلاك الطاقة لكل نمط قيادة عبر أنواع الطرق المختلفة
pivot_energy = df.pivot_table(
    index='Driving_Mode', 
    columns='Road_Type', 
    values='Energy_Consumption_kWh', 
    aggfunc='mean'
)
print("Average Energy Consumption by Driving Mode & Road Type:")
pivot_energy
```
